## Load environment variables from .env file (requires `python-dotenv`)

```python
from dotenv import load_dotenv
import os

load_dotenv()

llm_model = os.getenv("MODEL")
```

# Using Language Models

```python
from langchain_openai import ChatOpenAI

model = ChatOpenAI(
    model=llm_model,
    temperature=0.7,
    max_tokens=500,
    base_url="https://openrouter.ai/api/v1"
)

```

### เรียกใช้ message โดยตรงผ่าน `.invoke` method


LangChain also supports chat model inputs via strings or OpenAI format. The following are equivalent:

    model.invoke("Hello")

    model.invoke([{"role": "user", "content": "Hello"}])

    model.invoke([HumanMessage("Hello")])

```python
from langchain_core.messages import HumanMessage, SystemMessage

messages = [
    SystemMessage(content="Translate the following from Englist into Thai"),
    HumanMessage(content="Hello")
]

model.invoke(messages)
```

```python
# แสดงเฉพาะคำตอบ (content)

response = model.invoke(messages)
response.content
```

### Streaming

stream individual tokens from a chat model:

```python
for token in model.stream(messages):
    print(token.content, end="|")
```

# Prompt templates

Prompt templates are a concept in LangChain designed to assist with this transformation. They take in raw user input and return data (a prompt) that is ready to pass into a language model.

Note: that ChatPromptTemplate supports multiple message roles in a single template.

```python
from langchain_core.prompts import ChatPromptTemplate

system_template = "Translate the following from Thai into {language}"

prompt_template = ChatPromptTemplate.from_messages([
    ("system", system_template),
    ("user", "{text}")
])
```

it returns a ChatPromptValue that consists of two messages.

```python
prompt = prompt_template.invoke({"language":"Japan","text":"สวัสดี"})
prompt
```

If we want to access the messages directly we do `to_messages()`:

```python
prompt.to_messages()
```

Finally, we can invoke the chat model on the formatted prompt:

```python
response = model.invoke(prompt)
print(response.content)
```
